```python
import numpy as np
from Bio import SeqIO
import random
import numpy as np
import gzip
```

```python
fasta_file = "/home/hhousseinho/MIAT/Documents/Srna_Classification/Data/TRNA_RFAM.fasta.gz"

with gzip.open(fasta_file, "rt") as handle:
    parse = list(SeqIO.parse(handle, "fasta"))
```

```python
print(parse[0].description)
print(parse[0].seq)
```

```python
import numpy as np
n=4385

description = np.array([str(k.description) for k in parse]).tolist()
sequence = np.array([str(k.seq) for k in parse]).tolist()
```

```python
description_finale = np.random.choice(description,n,replace=False)
print(len(description_finale))
description_finale
```

```python
sequence_finale = np.random.choice(sequence,n,replace=False)
print(len(sequence_finale))
sequence_finale
```

```python
ofile = open("/home/hhousseinho/MIAT/Documents/Srna_Classification/Data/TRNA_RFAM_Reducted.fasta", "w")

for i in range(len(sequence_finale)):
    ofile.write(">" + description_finale[i] + "\n" + sequence_finale[i] + "\n")

#do not forget to close it
ofile.close()
```
