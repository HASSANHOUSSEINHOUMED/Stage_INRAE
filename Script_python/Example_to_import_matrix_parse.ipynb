```python
import numpy as np
import os 
from scipy.sparse import coo_matrix
import seaborn as sns
import pandas as pd
```

```python
# TRNA_RFAM

# folder path
your_folder_path='/home/hhousseinho/MIAT/Documents/Srna_Classification/RNAfold/TRNA_RFAM_Reducted'

# list of files
list_of_files=os.listdir(your_folder_path)

path_of_matrix=[]

for your_file_name in list_of_files:
    if 'dp.ps' in your_file_name:
        path_of_matrix.append(your_folder_path + '/' + your_file_name)

print(path_of_matrix)
```

```python
with open(path_of_matrix[0],'r') as f:
    lines = f.readlines()
    
modified = []

for line in lines:
    modified.append(line.strip())

print(modified)

index_debut = modified.index('%start of base pair probability data')
index_fin = modified.index('showpage')
```

```python
list1 = modified[index_debut+1:index_fin]
list1
```

```python
p = []

for k in range(len(list1)):
    l=list1[k].split()
    p.append(l)
    
p
```

```python
p=np.array(p)
p=p[p[:,3]=="ubox"]
print(p)

p=p[:,0:3]
p=p.astype("float")
p
```

```python
dat2 = pd.DataFrame(p)
dat2
```

```python
dat2.columns=['i','j','x']
```

```python
row  = dat2["i"].astype("int")-1

col  = dat2["j"].astype("int")-1

data = dat2["x"]

maxIndex=max(col)
print(maxIndex)

matrix_parse = coo_matrix((data, (row, col)), shape=(173, 173))
print(matrix_parse)
```

```python
matrix_parse.toarray()
```

```python
sns.heatmap(matrix_parse.toarray()[0:maxIndex,0:maxIndex])
```
